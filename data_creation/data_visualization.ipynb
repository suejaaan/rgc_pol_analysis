```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import numpy as np
```

```python
file_path=os.path.expanduser("~/Downloads/2024-01-29_17h00m43s-base-PolySignal.csv")
data=pd.read_csv(file_path)
data
```

```python
data_signal = data.iloc[:, 1:]

# Plot the second row (index 1)
plt.figure(figsize=(12, 5))
plt.plot(range(data_signal.shape[1]), data_signal.iloc[1].values)
plt.title("")
plt.xlabel("Sample Index")
plt.ylabel("Signal Value")
plt.tight_layout()
plt.show()
```

```python
#df = data.iloc[5000, :-1]  # Ignore the last column

# Extract the polarization value (second last column)
polarization_value = data.iloc[data['P'].idxmax(), -2]
df = data.iloc[data['P'].idxmax(), :-1]  
x_values = np.linspace(30.88, 34.48, 500)
plt.figure(figsize=(10, 5))
plt.plot(x_values,df[:500], marker='o', linestyle='-', color='r', markersize=3)
plt.title(f"Polarization: {polarization_value * 100:.2f} %")
plt.xlabel("Frequency (MHz)")
plt.ylabel("Signal (mV)")
plt.grid(False)
#plt.savefig("noise.png")
plt.show()
```
